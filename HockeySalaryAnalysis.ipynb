```python
import pandas as pd

df = pd.read_csv("player_dataset.csv.xls")

df.isnull().sum()
df.duplicated().sum()
df.dtypes
df = df.drop(columns=['CLAUSE'])

df['SALARY'] = df['SALARY'].astype(str)
df['SALARY'] = df['SALARY'].str.replace('$', '')
df['SALARY'] = df['SALARY'].str.replace(',', '')
df['SALARY'] = df['SALARY'].astype(int)

df['SALARY'] = pd.to_numeric(df['SALARY'], errors='coerce')


df[['SALARY', 'P', 'G', 'A', 'GP']].describe()
df[['SALARY', 'P/GP']].corr()

import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(df['SALARY'], df['P/GP'])
plt.xlabel("Salary (USD)")
plt.ylabel("Points Per Game")
plt.title("Relationship Between Salary and Points Per Game")
plt.show()

top_paid = df.sort_values('SALARY', ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.bar(top_paid['name'], top_paid['P/GP'])
plt.xticks(rotation=45)
plt.xlabel("Player")
plt.ylabel("Points Per Game")
plt.title("Points Per Game of Top 10 Highest Paid Players")
plt.show()
```
